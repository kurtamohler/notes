# Distributions

How are probability distributions generated in PyTorch?

## CDF

Probability distributions can be defined by a [cumulative distribution function](https://en.wikipedia.org/wiki/Cumulative_distribution_function) (CDF), $F(x)$. For any given $x \in \mathbb R$, $F(x)$ is a value between $0$ and $1$ which indicates the proportion of the values in the distribution which are less than or equal to $x$. So $F(x)$ is a monotonically increasing function.

If $F(x_0) = 0$, then all values in the distribution are $x \ge x_0$. If $F(x_0) = 1$, then all values in the distribution are $x \le x_0$.

For example, the CDF for the uniform distribution between 0 and 1 is:

$$
U_{[0, 1]}(x) = x
\quad\quad
0 \le x \le 1
$$

(To be totally accurate, we should also say that $U(x) = 0$ if $x < 0$ and $U(x) = 1$ if $x > 1$.)

We can see that this is correct by reasoning that at $x = 1/4$, one fourth of the distribution lies to the left, or at $x = 6/7$, six sevenths of the distribution lies be to the left, etc.
